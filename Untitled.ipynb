```python

import numpy as np
import pandas as pd
import matplotlib

matplotlib.use("Agg")

import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import json

# 1. Generate reproducible random data
rng = np.random.default_rng(6)

dates = pd.date_range(start="2021-01-01",end="2023-12-31",freq="D")

df = pd.DataFrame({"Date": dates})
day = df["Date"].dt.dayofyear.to_numpy()

# 2. Generate seasonal temperature
seasonal_temp = 15 + 10 * np.sin(2 * np.pi * (day - 110) / 365.25)
noise = np.zeros(len(df))

for i in range(1, len(df)):
    noise[i] = (0.7 * noise[i - 1]+ rng.normal(0, 2))

temperature = seasonal_temp + noise

# 3. Generate seasonal rainfall
rain_probability = (0.18+ 0.10 * (1 + np.cos(2 * np.pi * (day - 320) / 365.25)) / 2)

wet_day = rng.random(len(df)) < rain_probability

rainfall = np.where(wet_day,rng.gamma(shape=2, scale=2.5, size=len(df)),0.0)

# 4. Add 15 heavy-rain spells
cooling = np.zeros(len(df))

for _ in range(15):
    start = rng.integers(0, len(df) - 8)
    duration = rng.integers(3, 9)

    rainfall[start:start + duration] += rng.gamma(shape=2, scale=6, size=duration)

    cooling[start:start + duration] += 2

temperature -= cooling

df["Temperature_C"] = temperature
df["Rainfall_mm"] = rainfall

# 5. Create monthly climate summary
df["Month"] = df["Date"].dt.month

monthly = df.groupby("Month").agg(Mean_Temperature=("Temperature_C", "mean"),Min_Temperature=("Temperature_C", "min"),
    Max_Temperature=("Temperature_C", "max"),Monthly_Rainfall=("Rainfall_mm", "sum"))

monthly["Average_Monthly_Rainfall"] = (df.groupby(["Month", df["Date"].dt.year])["Rainfall_mm"].sum().groupby(level=0).mean())

# 6. Calculate the 30-day rolling mean

df["Rolling_30_Day"] = (df["Temperature_C"].rolling(window=30, center=True, min_periods=1).mean())
df["Monthly_Average"] = df.groupby("Month")["Temperature_C"].transform("mean")

df["Anomaly"] = (df["Temperature_C"] - df["Monthly_Average"])

# 7. Find the hottest, coldest and wettest months
hottest_month = monthly["Mean_Temperature"].idxmax()
coldest_month = monthly["Mean_Temperature"].idxmin()
wettest_month = monthly["Average_Monthly_Rainfall"].idxmax()

month_names = ["Jan", "Feb", "Mar", "Apr", "May", "Jun","Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

five_day_rain = df["Rainfall_mm"].rolling(5).sum()
heaviest_end = five_day_rain.idxmax()

# 8. Print the results
results = {"hottest_month": month_names[hottest_month - 1],"hottest_mean_C": round(monthly.loc[hottest_month, "Mean_Temperature"], 2),
    "coldest_month": month_names[coldest_month - 1],"coldest_mean_C": round(monthly.loc[coldest_month, "Mean_Temperature"], 2),"wettest_month": month_names[wettest_month - 1],
    "average_rainfall_in_wettest_month_mm": round(monthly.loc[wettest_month, "Average_Monthly_Rainfall"], 2 ),"annual_rainfall_average_mm": round(
    df.groupby(df["Date"].dt.year)["Rainfall_mm"].sum().mean(), 2),"heaviest_5_day_rainfall_mm": round(five_day_rain.max(), 2),"heaviest_5_day_period_ends": str(
    df.loc[heaviest_end, "Date"].date()),"largest_positive_anomaly_C": round(df["Anomaly"].max(), 2),"largest_negative_anomaly_C": round(df["Anomaly"].min(), 2),
    "days_over_4C_from_monthly_average_percent": round((df["Anomaly"].abs() > 4).mean() * 100, 2)}

print("WEATHER SEASONALITY REPORT")
print("-" * 35)

for key, value in results.items():
    print(f"{key}: {value}")

# 9. Save the dataset and summary
df.to_csv("weather.csv", index=False)

with open("results.json", "w") as file:
    json.dump(results, file, indent=4)

# 10. Draw four charts
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Chart 1: Daily temperature and rolling mean
axes[0, 0].plot(df["Date"], df["Temperature_C"],alpha=0.45, linewidth=0.7, label="Daily temperature")
axes[0, 0].plot(df["Date"], df["Rolling_30_Day"],linewidth=2, label="30-day rolling mean")
axes[0, 0].set_title("Daily Temperature and Rolling Mean")
axes[0, 0].set_ylabel("Temperature (°C)")
axes[0, 0].legend()

# Chart 2: Daily temperature anomalies
axes[0, 1].plot(df["Date"], df["Anomaly"], linewidth=0.7)
axes[0, 1].axhline(0, linestyle="--", color="black")
axes[0, 1].fill_between(df["Date"], df["Anomaly"], 0,where=(df["Anomaly"] >= 0), alpha=0.5)
axes[0, 1].fill_between(df["Date"], df["Anomaly"], 0,where=(df["Anomaly"] < 0), alpha=0.5)
axes[0, 1].set_title("Temperature Anomalies")
axes[0, 1].set_ylabel("Difference (°C)")

# Chart 3: Monthly average temperature
axes[1, 0].bar(month_names,monthly["Mean_Temperature"].to_numpy())
axes[1, 0].set_title("Average Temperature by Month")
axes[1, 0].set_ylabel("Temperature (°C)")

# Chart 4: Average rainfall by calendar month
axes[1, 1].bar(month_names,monthly["Average_Monthly_Rainfall"].to_numpy())
axes[1, 1].set_title("Average Monthly Rainfall")
axes[1, 1].set_ylabel("Rainfall (mm)")

# Format date axes
for ax in axes.flat[:2]:
    ax.xaxis.set_major_locator(mdates.MonthLocator(interval=4))
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%b %Y"))
    ax.tick_params(axis="x", labelrotation=30)

fig.suptitle("Weather Seasonality Analysis (2021–2023)")
fig.tight_layout()
fig.savefig("chart.png", dpi=120, bbox_inches="tight")

print("\nFiles saved: weather.csv, results.json, chart.png")
```
